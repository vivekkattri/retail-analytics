# Cohort Retention

```python
customers = pd.read_parquet(root / "data" / "processed" / "customers.parquet")
customers["Month"] = customers["InvoiceDate"].dt.to_period("M")
first_month=customers.groupby('Customer ID')["Month"].min()
first_month
```

```python
customers["CohortMonth"] = customers["Customer ID"].map(first_month)
customers.head(20)
```

```python
customers["CohortIndex"] = (customers["Month"] - customers["CohortMonth"]).apply(lambda x: x.n)
customers[["Month", "CohortMonth", "CohortIndex"]].head(20)
```

```python
counts = customers.groupby(["CohortMonth", "CohortIndex"])["Customer ID"].nunique()
counts
```

```python
cohort_table = counts.unstack(1)
cohort_table
```
